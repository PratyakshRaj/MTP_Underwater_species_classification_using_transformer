```python
import os, json
```

```python
root="downloads"
```

```python
sp_list=os.listdir(root)
species_to_idx={ fname:id for id,fname in enumerate(sp_list) }
```

```python
label_map={}
count=0
for sp in sp_list:
    file_list=os.path.join(root,sp)
    for aud_file in os.listdir(file_list):
        label_map[sp+"//"+aud_file]=species_to_idx[sp]
        count+=1
```

```python
with open("label_map.json","w") as f:
    json.dump(label_map,f,indent=2)
    
```
